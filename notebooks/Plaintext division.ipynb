```python
import math
import numpy as np
from random import randrange
import random
```

```python
bits = 128

a = random.getrandbits(bits)
b = random.getrandbits(bits)

reciprocal = (1 << (bits + b.bit_length())) // b
print(reciprocal)

result = (a * reciprocal)

result = result >> b.bit_length()
result = result >> bits


print("A:", a)
print("B:", b)
print("My res:", result)
print("Expect:", a // b)
```
