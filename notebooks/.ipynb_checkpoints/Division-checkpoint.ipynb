```python
import math
import numpy as np
import matplotlib.pyplot as plt
from random import randrange
import random
```

```python
# A power of two
bits = 128

a = random.getrandbits(bits)
b = random.getrandbits(bits)

LUT_BITS = 8
LUT_SIZE = 1 << LUT_BITS

lut = [
    (1 << (bits * 2)) //
    ((1 << (bits - 1)) + (i << (bits - 1 - LUT_BITS)))
    for i in range(LUT_SIZE)
]

b_hat = b << (bits - b.bit_length())

i = (b_hat >> (bits - 1 - LUT_BITS)) % LUT_SIZE

x = lut[i]

for _ in range(bits // LUT_BITS):
    x = (x * ((1 << (bits * 2 + 1)) - b_hat * x)) >> (bits * 2)

result = (a * x)
result = result >> (bits + b.bit_length())

print("***")
print("A:", a)
print("B:", b)
print("My res:", result)
print("Expect:", a // b)
```

## Mapping the LUT to LUT_BITS polynomials

```python
def get_bit(n, i):
    return (n >> i) & 1
```

```python
def polylut(bit = 0):
    N = LUT_SIZE
    seq = np.array([get_bit(lut[i], bit) for i in range(N)], dtype=float)
    c = np.fft.fft(seq)  

    def f(x):
        s = 0.0
        for k in range(N):
            ang = 2 * math.pi * k * x / N
            s += c[k].real * math.cos(ang) - c[k].imag * math.sin(ang)
        return s / N

    return f
```

```python
class Chebyshev:
    """
    Chebyshev(a, b, n, func)
    Given a function func, lower and upper limits of the interval [a,b],
    and maximum degree n, this class computes a Chebyshev approximation
    of the function.
    Method eval(x) yields the approximated function value.
    """
    def __init__(self, a, b, n, func):
        n = n + 1
        self.a = a
        self.b = b
        self.func = func

        bma = 0.5 * (b - a)
        bpa = 0.5 * (b + a)
        f = [func(math.cos(math.pi * (k + 0.5) / n) * bma + bpa) for k in range(n)]
        self.roots = f

        self.x = [math.cos(math.pi * (k + 0.5) / n) * bma + bpa for k in range(n)]

        fac = 2.0 / n
        self.c = [fac * sum([f[k] * math.cos(math.pi * j * (k + 0.5) / n)
                  for k in range(n)]) for j in range(n)]

    def eval(self, x):
        a,b = self.a, self.b
        #assert(a <= x <= b)
        y = (2.0 * x - a - b) * (1.0 / (b - a))
        y2 = 2.0 * y
        (d, dd) = (self.c[-1], 0)             # Special case first step for efficiency
        for cj in self.c[-2:0:-1]:            # Clenshaw's recurrence
            (d, dd) = (y2 * d - dd + cj, d)
        return y * d - dd + 0.5 * self.c[0]   # Last step is different
```

```python
# Notice: for large bit sizes (64, 128, ...), this might take a few minutes

polyluts = []

for bit in range(bits + 2):
    f = polylut(bit)
    
    c = Chebyshev(0, LUT_SIZE, 851, f)
    
    polyluts.append(c)
```

```python
# For instance, this polynomial represents the first bit of the LUT result

x = np.arange(0, LUT_SIZE, 1)
y = np.vectorize(polyluts[0].eval)(x)

plt.plot(x, y)

y_real = np.vectorize(polylut(0))(x)
plt.plot(x, y_real)


max(abs(y - y_real))
```

### Now evaluating the division using polynomials only

```python
a = random.getrandbits(bits)
b = random.getrandbits(bits)

LUT_BITS = 8
LUT_SIZE = 1 << LUT_BITS

b_hat = b << (bits - b.bit_length())

i = b_hat >> (bits - 1 - LUT_BITS)
i = i % 2**LUT_BITS

#####################################
###### Binary LUT evaluation ########
#####################################

binx = []

for j in range(bits * 2):
    if j >= len(polyluts):
        binx.append(0)
    else:
        binx.append(polyluts[j].eval(i))
        
    
polydec = ([round(x) for x in list(binx)][::-1])
x = int("".join(map(str, polydec)), 2)

#####################################
#####################################
#####################################

for _ in range(bits // LUT_BITS):
    x = (x * ((1 << (bits * 2 + 1)) - b_hat * x)) >> (bits * 2)

result = (a * x)
result = result >> (bits + b.bit_length())


print("A         :", a)
print("B         :", b)
print("My result :", result)
print("Expected  :", a // b)
```
